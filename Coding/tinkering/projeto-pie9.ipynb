## 🖼️ Treinamento do Modelo com o dataset e YOLO11

Configuração e execução do treinamento para detecção de defeitos em vidros laminados usando YOLOv11

🧪 Install Ultralytics

```python
# Instala a biblioteca Ultralytics (YOLO11)
!pip install ultralytics
```

📚 Importações e Configurações

```python
import os
import time
from datetime import datetime
from ultralytics import YOLO
import ultralytics

# Configurações do treinamento e predição
MODEL_PATH = 'yolo11m.pt'  # Caminho para o modelo inicial
DATA_YAML = '/kaggle/input/glass-defect/config.yaml'  # Caminho para o YAML do dataset

# 📁 Diretório base do dataset (mesmo nível do notebook)
dataset_dir = '/kaggle/input/glass-defect/dataset/dataset'

# Hiperparâmetros
EPOCHS = 10
PATIENCE = 200
CONF_TRAIN = 0.3
CONF_PREDICT = 0.6
SINGLE_CLASS = True

# Caminhos para imagens e resultados
IMAGE_TO_PREDICT_DIR = os.path.join(dataset_dir, 'images', 'test')
RESULT_DIR = os.path.join('/kaggle/working/runs', 'results')
```

🧪 Verificar ambiente

```python
print("🔍 Verificando ambiente YOLO...")
ultralytics.checks()
```

🧠 Carregar o modelo

```python
print(f"📦 Carregando modelo: {MODEL_PATH}")
model = YOLO(MODEL_PATH)
```

🚀 Treinamento do modelo

```python
print("🚀 Iniciando treinamento do modelo...")
train_results = model.train(
    data=DATA_YAML,
    epochs=EPOCHS,
    patience=PATIENCE,
    val=True,
    batch=12,
    single_cls=SINGLE_CLASS,
    lr0=0.002,
    device='cuda',
)
```

📈 Avaliação do modelo

```python
print("📈 Avaliando modelo no conjunto de validação...")
val_results = model.val(conf=CONF_TRAIN, save_json=True)
```

🖼️ Coletar imagens para predição

```python
print(f"🖼️ Buscando imagens em: {IMAGE_TO_PREDICT_DIR}")
image_extensions = ('.jpg', '.jpeg', '.png', '.bmp')
image_paths = [
    os.path.join(IMAGE_TO_PREDICT_DIR, f)
    for f in os.listdir(IMAGE_TO_PREDICT_DIR)
    if f.lower().endswith(image_extensions)
]

if not image_paths:
    print(f"❌ Nenhuma imagem encontrada em {IMAGE_TO_PREDICT_DIR}")
else:
    print(f"🔍 {len(image_paths)} imagens encontradas para predição.")
```

🔮 Fazer predições

```python
# Executa predição se houver imagens
if image_paths:
    pred_results = model.predict(source=image_paths, conf=CONF_PREDICT)
```

💾 Salvar os resultados

```python
# Garante que a pasta de resultados existe
os.makedirs(RESULT_DIR, exist_ok=True)

# Salva os resultados com timestamp
for i, result in enumerate(pred_results):
    timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
    filename = f"result_{i+1}_{timestamp}.jpg"
    result_path = os.path.join(RESULT_DIR, filename)
    result.save(filename=result_path)
    print(f"💾 Resultado salvo em: {result_path}")
```

```python
print("✅ Execução finalizada com sucesso!")
```

## 🎯 Inferência com YOLO11 em vídeo da câmera

Detecta defeitos com o Modelo treinado em YOLO11 a partir de uma URL de câmera.

📸

```python
# 📦 Importar bibliotecas e definir constantes básicas
import cv2
import time
import threading
import numpy as np
from ultralytics import YOLO
import os

#MODEL_PATH = 'best.pt'
CAMERA_URL = 'http://10.109.61.34:8080/video'  # URL da câmera IP

model = YOLO(MODEL_PATH)

running = True
annotated_frame = None
frame_for_inference = None
lock = threading.Lock()
```

```python
def inference_thread():
    global frame_for_inference, annotated_frame, running

    while running:
        with lock:
            if frame_for_inference is not None:
                frame = frame_for_inference.copy()
                frame_for_inference = None
            else:
                frame = None

        if frame is not None:
            results = model.predict(source=frame, conf=0.5, verbose=False)[0]

            annotated = frame.copy()
            for box in results.boxes:
                x1, y1, x2, y2 = map(int, box.xyxy[0])
                cv2.rectangle(annotated, (x1, y1), (x2, y2), (0, 0, 255), 2)
                cv2.putText(annotated, "DEFECT", (x1, y1 - 10),
                            cv2.FONT_HERSHEY_SIMPLEX, 0.9, (0, 0, 255), 2)

            with lock:
                annotated_frame = annotated

        time.sleep(0.01)
```

```python
cap = cv2.VideoCapture(CAMERA_URL)
if not cap.isOpened():
    raise RuntimeError("❌ Erro ao acessar a câmera IP.")

thread = threading.Thread(target=inference_thread)
thread.start()

print("✅ Câmera conectada. Pressione Ctrl+C para parar.")

try:
    while True:
        ret, frame = cap.read()
        if not ret:
            print("⚠️ Falha ao capturar frame.")
            break

        with lock:
            frame_for_inference = frame.copy()
            display_frame = annotated_frame if annotated_frame is not None else frame

        # Converte BGR para RGB para matplotlib
        display_frame_rgb = cv2.cvtColor(display_frame, cv2.COLOR_BGR2RGB)

        clear_output(wait=True)
        plt.imshow(display_frame_rgb)
        plt.axis('off')
        plt.show()

        time.sleep(0.1)  # Ajusta para não saturar GPU/CPU

except KeyboardInterrupt:
    print("🛑 Interrompido manualmente.")

finally:
    running = False
    thread.join()
    cap.release()
    plt.close('all')
    print("👋 Finalizado.")
```
